## PE2 (3.5 pts) - Simulación moneda trucada

```
# SOLUCIÓN PE2 - PARTE 1: Cálculo
print("\n" + "=" * 80)
print("PE2 - MONEDA TRUCADA: CARA 70%")
print("=" * 80)

# Parámetros de simulación
np.random.seed(42)  # Para reproducibilidad
num_lanzamientos = 50
prob_cara = 0.7

# Simular lanzamientos
# 1 = cara, 0 = cruz
lanzamientos = np.random.binomial(n=1, p=prob_cara, size=num_lanzamientos)

print(f"\n🎲 Simulación de {num_lanzamientos} lanzamientos:")
print(f"\nResultados (1=cara, 0=cruz):")
print(lanzamientos)

# Análisis
total_caras = lanzamientos.sum()
total_cruces = num_lanzamientos - total_caras
porcentaje_caras = (total_caras / num_lanzamientos) * 100

print(f"\n📊 Resultados:")
print(f"   Total de caras: {total_caras}")
print(f"   Total de cruces: {total_cruces}")
print(f"   Porcentaje de caras: {porcentaje_caras:.2f}%")
print(f"   Porcentaje esperado: 70%")
print(f"   Diferencia: {abs(porcentaje_caras - 70):.2f}%")
```

```
# SOLUCIÓN PE2 - PARTE 2: Gráfico

# Calcular porcentaje acumulado de caras
caras_acumuladas = np.cumsum(lanzamientos)
porcentaje_acumulado = (caras_acumuladas / np.arange(1, num_lanzamientos + 1)) * 100

# Crear gráfico de línea
plt.figure(figsize=(12, 6))

# Gráfico de línea con puntos
plt.plot(range(1, num_lanzamientos + 1), porcentaje_acumulado,
         marker='o', linestyle='-', linewidth=2, markersize=4,
         color='#2E86AB', label='Porcentaje de caras')

# Línea de referencia en 70%
plt.axhline(y=70, color='red', linestyle='--', linewidth=2,
            label='Probabilidad esperada (70%)')

# Etiquetas y título
plt.xlabel('Número de lanzamiento', fontsize=12, fontweight='bold')
plt.ylabel('Porcentaje de caras (%)', fontsize=12, fontweight='bold')
plt.title('Moneda Trucada: Convergencia a 70% caras en 50 lanzamientos',
          fontsize=14, fontweight='bold')

plt.grid(True, alpha=0.3)
plt.legend(fontsize=11)
plt.ylim(0, 100)
plt.xlim(0, num_lanzamientos)

plt.tight_layout()
plt.savefig('grafico_pe2_moneda.png', dpi=100, bbox_inches='tight')
print("\n✅ Gráfico guardado como 'grafico_pe2_moneda.png'")
plt.show()
```

## PE3 (7 pts) - Primos de Mersenne menores que 100

```
# SOLUCIÓN PE3
print("\n" + "=" * 80)
print("PE3 - PRIMOS DE MERSENNE MENORES QUE 100")
print("=" * 80)

def es_primo(n):
    """
    Verifica si un número es primo.
    
    Criterios:
    - Números < 2: no son primos
    - 2: es primo
    - Pares: no son primos
    - Divisibles por n en rango [3, √n]: no son primos
    """
    if n < 2:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    
    # Verificar divisibilidad hasta √n
    for i in range(3, int(n**0.5) + 1, 2):
        if n % i == 0:
            return False
    return True

def encontrar_primos_mersenne(limite):
    """
    Encuentra todos los primos de Mersenne menores que 'limite'.
    
    Un primo de Mersenne cumple:
    1. Es un número primo
    2. Se expresa como 2^p - 1, donde p también es primo
    
    Ejemplos:
    - 3 = 2^2 - 1 (p=2, primo)
    - 7 = 2^3 - 1 (p=3, primo)
    - 31 = 2^5 - 1 (p=5, primo)
    """
    primos_mersenne = []
    
    # Búsqueda: para cada número primo candidato
    for candidato in range(2, limite):
        if not es_primo(candidato):
            continue  # Solo considerar números primos
        
        # Ahora verificar si es de Mersenne
        # Buscar p primo tal que 2^p - 1 = candidato
        p = 2
        encontrado = False
        
        while True:
            valor_mersenne = (2 ** p) - 1
            
            # Si ya pasamos el candidato, no será Mersenne
            if valor_mersenne > candidato:
                break
            
            # Si encontramos coincidencia Y p es primo
            if valor_mersenne == candidato and es_primo(p):
                primos_mersenne.append({
                    'numero': candidato,
                    'formula': f'2^{p} - 1',
                    'p': p
                })
                encontrado = True
                break
            
            # Siguiente número primo
            p += 1
            while not es_primo(p):
                p += 1
    
    return primos_mersenne

# Ejecutar búsqueda
print("\n🔍 Buscando primos de Mersenne menores a 100...\n")

resultado = encontrar_primos_mersenne(100)

if resultado:
    print(f"✅ Se encontraron {len(resultado)} primos de Mersenne:\n")
    for primo in resultado:
        print(f"   {primo['numero']:3d} = {primo['formula']:8s} (donde p={primo['p']} es primo)")
        # Verificación
        valor = (2 ** primo['p']) - 1
        print(f"        Verificación: 2^{primo['p']} - 1 = {valor} ✓")
else:
    print("❌ No se encontraron primos de Mersenne")

# Tabla resumen
print(f"\n📋 TABLA RESUMEN:")
print(f"   Primo | Fórmula   | p (primo) | Verificación")
print(f"   ------|-----------|-----------|------------")
for primo in resultado:
    p = primo['p']
    valor = (2 ** p) - 1
    print(f"   {primo['numero']:4d}  | {primo['formula']:9s} | {p:9d} | 2^{p}-1 = {valor}")
```
